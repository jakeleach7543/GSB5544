1\. The corpus is the body column in the [data set](https://raw.githubusercontent.com/kevindavisross/data301/refs/heads/main/data/enron_email.csv); read this column in to create the corpus.


```python
emails = pd.read_csv(
    "https://raw.githubusercontent.com/kevindavisross/data301/refs/heads/main/data/enron_email.csv")

corpus = emails["body"]

corpus.head()
```

```python
corpus = emails["body"].fillna("")
corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vec = CountVectorizer()
tf = vec.fit_transform(corpus)
tf_df = pd.DataFrame(tf.toarray(), columns=vec.get_feature_names_out())
```

```python
tf_df
```

4\. Explain what a document frequency (DF) is, and the idea of an IDF. Why might we use TF-IDF rather than just TF? This question is asking about concepts rather than computations.


**A document frequency is the amount of times a given Int64 value appears. IDF gives larger weights to rare words and less weight to more common words. TF-IDF is a combination where the value is high when a word appears a lot in a given email and is rare throughout the entire corpus**

5\. Use `TfidfVectorizer` to create the TF matrix, and then convert it to a pandas data frame with appropriate column names. You can just use the defaults in `TfidfVectorizer`. Hint: this should pretty much be a straight copy of code from the reading, but make sure you understand the process behind the code.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vec = TfidfVectorizer()
tfidf = tfidf_vec.fit_transform(corpus)
tfidf_df = pd.DataFrame(tfidf.toarray(), columns=tfidf_vec.get_feature_names_out())
```

```python
tfidf_df
```

8\. Consider just one email with unknown spam status, with `loc` 1980. Compute the cosine distance between the TF representation of this email and the TF representation of each of the emails with known spam status.

```python
from sklearn.metrics.pairwise import cosine_distances

# Emails with known spam status
known_mask = emails["spam"].notna()
known_idx = emails.index[known_mask]

# Email with unknown spam status
target_idx = 1980

distances = cosine_distances(
    tf_df.loc[[target_idx]],
    tf_df.loc[known_idx]).ravel()

results = pd.DataFrame({
    "loc": known_idx,
    "spam": emails.loc[known_idx, "spam"].values,
    "distance": distances})

results.head()
```

9. Sort the results of the previous part to examine the emails with known spam status that are most similar to the unknown email. If you had to determine if the unknown email was spam or not, which would you pick and why?

```python
results.sort_values("distance").head(10)
```

Since the emails with the known lowest cosine distance are spam, I would argue email 1980 is spam. Email 1980 seems to have a similar word frequency pattern to predetermined spam emails

10\. Repeat the two previous parts using TF-IDF instead of TF. Do the results change?

```python
from sklearn.metrics.pairwise import cosine_distances

# Known spam statuses
known_idx = emails.index[emails["spam"].notna()]

# Unknown email we are classifying
target_idx = 1980

distances_tfidf = cosine_distances(
    tfidf_df.loc[[target_idx]],
    tfidf_df.loc[known_idx]).ravel()

results_tfidf = pd.DataFrame({
    "loc": known_idx,
    "spam": emails.loc[known_idx, "spam"].values,
    "distance": distances_tfidf})

results_tfidf = results_tfidf.sort_values("distance")

results_tfidf.head(10)
```

Using TF-IDF does not change our results significantly. Our results show that four of the five emails with the smallest cosine distances are spam, so we still maintain that email 1980 is spam

11\. Now proceed in a similar manner to "guess" the spam status of each of the unknown emails.

```python
# Emails where spam status is known
known_idx = emails.index[emails["spam"].notna()]

# The 10 emails where spam status is unknown
unknown_idx = emails.index[emails["spam"].isna()]

guesses = []

# Repeat the process for each unknown email
for target_idx in unknown_idx:

    distances = cosine_distances(
        tfidf_df.loc[[target_idx]],
        tfidf_df.loc[known_idx]).ravel()

    results = pd.DataFrame({
        "loc": known_idx,
        "spam": emails.loc[known_idx, "spam"].values,
        "distance": distances})

    closest5 = results.sort_values("distance").head(5)

    
    prediction = int(closest5["spam"].mean() > 0.5)

    guesses.append({
        "loc": target_idx,
        "predicted_spam": prediction})

predictions = pd.DataFrame(guesses)

predictions
```

Emails 1980, 1985-1988 are predicted to be spam. Emails 1981-1984, 1989 are not predicted as spam.
