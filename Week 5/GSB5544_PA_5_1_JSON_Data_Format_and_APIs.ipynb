Now you will work with the [TVMaze API](http://www.tvmaze.com/api) from the reading. Use the API to request JSON data that you can use to answer the following questions.

```python
import requests

response = requests.get(
    "https://api.tvmaze.com/schedule",
    params={
        "country": "US",
        "date": "2018-02-04"
    }
)

data_schedule = response.json()
```

```python
df_schedule = pd.json_normalize(data_schedule)

df_schedule.head()
```

```python
df_schedule.sort_values("runtime", ascending=False)[["show.name","runtime","airtime"]].head(2)
```

2\. Among all shows that aired in the U.S. on Feburary 4, 2018, which non-voice actors appeared on more than one show? Note: some people are credited with multiple rows on the same show and thus appear as multiple rows in the data frame.

Hint: You will need to write a for loop to make multiple requests. Use the "show.id" from the data from part 1, and use the ["Shows" endpoint](http://www.tvmaze.com/api#show-cast) to get the cast of each show. Don't forget to stagger your requests, or you will be blocked by the website!

```python
show_ids = df_schedule["show.id"].unique()

show_ids
```

```python
## https://api.tvmaze.com/shows/1/cast
import time


cast_records = []
for show_id in show_ids:
    response = requests.get(f"https://api.tvmaze.com/shows/{show_id}/cast")
    if response.status_code != 200:
        print("Skipped", show_id, "State error", response.status_code)
        continue 
    members = response.json()
    for member in members:
        member["show_id"] = int(show_id)
    cast_records.extend(members)
    time.sleep(0.5)
```

```python
import time

cast = []

for show_id in show_ids:

    response = requests.get(
        f"https://api.tvmaze.com/shows/{show_id}/cast"
    )

    members = response.json()

    for m in members:
        m["show_id"] = show_id

    cast.extend(members)

    time.sleep(0.5)
```

```python
df_cast = pd.json_normalize(cast)

df_cast.head()
```

```python
df_nonvoice = df_cast[
    df_cast["voice"] == False]
```

```python
# Unique actors only
actor_shows = df_nonvoice[
    ["person.name", "show_id"]].drop_duplicates()
```

```python
actor_counts = (
    actor_shows
    .groupby("person.name")
    .size()
    .sort_values(ascending=False))

actor_counts
```

```python
actor_counts[
    actor_counts > 1]
```

```python
# after the loop
df_cast = pd.json_normalize(cast)

df_nonvoice = df_cast[
    df_cast["voice"] == False
]

actor_shows = df_nonvoice[
    ["person.name", "show_id"]
].drop_duplicates()

actor_counts = (
    actor_shows
    .groupby("person.name")
    .size()
    .sort_values(ascending=False)
)

actor_counts[
    actor_counts > 1
]
```

But note that in the recipes/list endpoint there are two keys: count and results. We only want the results so try instead `.extend(response.json()).get("results")`.

```python
import time

avocado_recipes = []

for start in range(0, 120, 40):

    querystring = {
        "q": "avocado",
        "size": "40",
        "from": str(start)
    }

    response = requests.get(
        url,
        headers=headers,
        params=querystring
    )

    avocado_recipes.extend(
        response.json().get("results")
    )

    time.sleep(0.5)

len(avocado_recipes)
```

```python
# run full loop for all 444 recipes
total_avocado = response.json()["count"]

avocado_recipes = []

for start in range(0, total_avocado, 40):

    querystring = {
        "q": "avocado",
        "size": "40",
        "from": str(start)
    }

    response = requests.get(
        url,
        headers=headers,
        params=querystring
    )

    avocado_recipes.extend(
        response.json().get("results")
    )

    time.sleep(0.5)

len(avocado_recipes)
```

4\. For the recipes containing avocado, compute the proportion of reviews that are positive. For the avocado recipes with over 500 reviews, which one has the highest proportion of positive reviews?


The recipe with the highest proportion of positive reviews is the grilled salmon

```python
df_avocado["total_reviews"] = (
    df_avocado["user_ratings.count_positive"]
    + df_avocado["user_ratings.count_negative"])

df_avocado["positive_proportion"] = (
    df_avocado["user_ratings.count_positive"]
    / df_avocado["total_reviews"])

df_avocado[
    df_avocado["total_reviews"] > 500][
    ["name", "total_reviews", "positive_proportion"]].sort_values(
    "positive_proportion",
    ascending=False).head(1)
```
