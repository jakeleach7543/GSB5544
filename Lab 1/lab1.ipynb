# Plot 1

```python
q1 = pd.read_csv("labs/Lab 1/Data/q1data.csv")

q1.head()
```

```python
q1.columns
```

1. For plot 1 aesthetics, life expectancy (life_exp) is tied to the y-axis, income (income) to the x-axis, population (population) to size, world region (four_regions) to color, with each point's shape being circular

2.

```python
plot1 = (
    ggplot(
        q1_plot,
        aes(
            x="income",
            y="life_exp",
            size="population",
            color="four_regions"))
    + geom_point(alpha=0.8)
    + scale_x_log10(
        breaks=[500, 1000, 2000, 4000, 8000, 16000, 32000, 64000],
        labels=lambda x: [
            "500", "1k", "2k", "4k", "8k", "16k", "32k", "64k"
        ][:len(x)])
    + scale_size_area(max_size=15)
    + labs(
        x="Income",
        y="Life expectancy",
        color="World Regions",
        size="Population")
    + theme_minimal())
```

```python
plot1
```

3.

```python
plot1_alt = (
    ggplot(
        q1_plot,
        aes(
            x="income",
            y="life_exp",
            color="four_regions",
            group="four_regions"))
    + geom_line()
    + scale_x_log10(
        breaks=[500, 1000, 2000, 4000, 8000, 16000, 32000, 64000],
        labels=lambda x: [
            "500", "1k", "2k", "4k", "8k", "16k", "32k", "64k"
        ][:len(x)])
    + labs(
        x="Income",
        y="Life expectancy",
        color="World Regions")
    + theme_minimal())
```

```python
plot1_alt
```

3. 
Besides geom_point, geom_line can somewhat visualize income and life expectancy, although less effective. Using lines to connect countries almost implies a continuous relationship between data points that may not be accurate. Using geom_point is better because each country can represent a unique observation a lot easier visually

# Plot 2

```python
q2 = pd.read_csv("labs/Lab 1/Data/q2data.csv")

q2.head()
q2.columns
```

1. For plot 2 aesthetics, imports (imports) are tied to the y-axis, exports (exports) to the x-axis, energy use (energy) to size, world region (four_regions) to color, with each point's shape being circular.

2.

```python
plot2 = (
    ggplot(
        q2_plot,
        aes(
            x="exports",
            y="imports",
            size="energy",
            color="four_regions"))
    + geom_point(alpha=0.8)
    + labs(
        x="Exports",
        y="Imports",
        color="World Regions",
        size="Energy Use")
    + theme_minimal())
```

```python
# recommended error fix
q2["imports"] = pd.to_numeric(q2["imports"], errors="coerce")
q2["exports"] = pd.to_numeric(q2["exports"], errors="coerce")
q2["energy"] = pd.to_numeric(q2["energy"], errors="coerce")
```

```python
# recommended error fix
q2_plot = q2.dropna(
    subset=["imports", "exports", "energy", "four_regions"]
)
```

```python
plot2
```

3.

```python
plot2_alt = (
    ggplot(
        q2_plot,
        aes(
            x="exports",
            y="imports",
            color="four_regions",
            group="four_regions"))
    + geom_line()
    + labs(
        x="Exports",
        y="Imports",
        color="World Regions")
    + theme_minimal())
```

```python

plot2_alt
```

3.
Geom_line can show the relationship between exports and imports, however, it's still not as clear as geom_point. Similar to plot 1, connecting the countries with lines makes it look continuous when in reality these countries are individual observations. Using a plot with points is better because it keeps the countries more distinct.

# Plot 3

```python
q3 = pd.read_csv("labs/Lab 1/Data/q3data.csv")

q3.head()
q3.columns
```

1. 
For plot 3 aesthetics, individuals using the internet (internet_users) is tied to the x-axis, GDP per capita (gdp) to the y-axis, income (income) to size, world region (four_regions) to color, with each point’s shape being circular.

2.

```python
q3_plot = q3.dropna(
    subset=["internet_users", "gdp", "income", "four_regions"])
```

```python
plot3 = (
    ggplot(
        q3_plot,
        aes(
            x="internet_users",
            y="gdp",
            size="income",
            color="four_regions"))
    + geom_point(alpha=0.8)
    + labs(
        x="Individuals Using the Internet",
        y="GDP per Capita",
        color="World Regions",
        size="Income")
    + theme_minimal())
```

```python
plot3
```

3.

```python
plot3_alt = (
    ggplot(
        q3_plot,
        aes(
            x="internet_users",
            y="gdp",
            color="four_regions"))
    + geom_smooth(se=False)
    + labs(
        x="Individuals Using the Internet",
        y="GDP per Capita",
        color="World Regions")
    + theme_minimal())
```

```python
plot3_alt
```

3.
This time I tried geom_smooth. It's useful for showing the overall relationship between internet usage and GDP per capita across regions, but it hides the individual country observations. geom_point is better for this dataset because it preserves each country as a unique data point.
